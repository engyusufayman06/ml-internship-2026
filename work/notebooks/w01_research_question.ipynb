**Lane:** Refresh / Content Opportunity Scoring

The goal is to prioritize pages for human review, not to automate content decisions.

## 1. My lane and why

I chose **Refresh / Content Opportunity Scoring** because the starter data contains observable search and content signals such as impressions, clicks, CTR, average position, content age, word count, and engagement. These signals make it possible to test whether a ranked review queue can focus limited editorial attention on pages that may need investigation.

## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should be reviewed first for a potential content refresh or other improvement?

**Decision:** assign higher review priority to pages with stronger evidence of opportunity or risk.

**Who acts:** a content/SEO reviewer.

**Action:** review, refresh, expand, monitor, or leave unchanged after human inspection.

**Wrong-call cost:** false positives consume review capacity; false negatives can cause useful opportunities to be missed.

## 3. Quick look at the data

The following cell records real starter-snapshot facts that motivate the project.

```python
from pathlib import Path
import pandas as pd

candidates = [Path('data/raw/content_refresh_anonymized.csv'), Path('../../data/raw/content_refresh_anonymized.csv')]
data_path = next((p for p in candidates if p.exists()), None)
if data_path is None:
    raise FileNotFoundError('Starter dataset not found.')
df = pd.read_csv(data_path)
print('Number of pages:', len(df))
print('Number of columns:', df.shape[1])
print('Declining proxy rate:', round(df['trend_direction'].astype(str).str.lower().eq('down').mean(), 3))
print('search_volume vs impressions_90d correlation:', round(df['search_volume'].corr(df['impressions_90d']), 3))
```
